## Загрузка результатов сканирования

Читаем CSV-файл, содержащий все прогоны параметрического сканирования.
Файл содержит следующие колонки:
- `beta` — коэффициент заразности
- `peak` — пиковая доля инфицированных
- `final_inf` — конечная доля инфицированных
- `final_rec` — конечная доля выздоровевших
- `deaths` — число умерших
- `seed` — зерно случайных чисел (для повторных прогонов)

```julia
df = CSV.read(datadir("beta_scan_all.csv"), DataFrame)
```

## Создание составного графика

Строим три вертикальных подграфика, каждый из которых показывает зависимость
одного эпидемического показателя от коэффициента заразности β.

### Панель 1: Динамика инфицированных

Отображает пиковую и конечную долю инфицированных.
- **Пик** (peak) — максимальная доля одновременно болеющих
- **Конечная доля** (final_inf) — общая доля переболевших к концу эпидемии

```julia
p1 = plot(
    df.beta,
    df.peak,
    label = "Пик",
    xlabel = "β",
    ylabel = "Доля инфицированных",
    linewidth = 2,
    marker = :circle,
)

plot!(
    p1,
    df.beta,
    df.final_inf,
    label = "Конечная",
    linewidth = 2,
    marker = :square,
)
```

### Панель 2: Число умерших

Показывает абсолютное количество смертей в зависимости от заразности.
Это важный показатель для оценки нагрузки на систему здравоохранения
и социальных последствий эпидемии.

```julia
p2 = plot(
    df.beta,
    df.deaths,
    xlabel = "β",
    ylabel = "Число умерших",
    linewidth = 2,
    marker = :diamond,
    legend = false,
)
```

### Панель 3: Доля выздоровевших

Отображает долю популяции, приобретшей иммунитет к концу эпидемии.
Этот показатель связан с порогом коллективного иммунитета.

```julia
p3 = plot(
    df.beta,
    df.final_rec,
    xlabel = "β",
    ylabel = "Доля выздоровевших",
    linewidth = 2,
    marker = :circle,
    legend = false,
)
```

## Объединение графиков

Размещаем три панели вертикально (3 строки, 1 столбец) с общим размером 800×900 пикселей.

```julia
plot(p1, p2, p3, layout = (3, 1), size = (800, 900))
```

Сохраняем составной график в каталог `plots/`

```julia
savefig(plotsdir("comprehensive_analysis.png"))
```

## Анализ полученных графиков

Составной график позволяет визуально оценить:

1. **Порог распространения** — минимальное значение β, при котором пик
   заболеваемости превышает фоновый уровень (обычно 5-10%)

2. **Нелинейность зависимости** — как быстро растут показатели с увеличением β

3. **Соотношение пика и конечной доли** — при высоких β пик может значительно
   превышать конечную долю, указывая на перегрузку системы

4. **Зависимость смертности** — растёт ли смертность линейно или ускоряется
   из-за коллапса системы здравоохранения

5. **Порог коллективного иммунитета** — значение β, при котором конечная доля
   выздоровевших достигает теоретического порога 1 - 1/R₀

Этот комплексный анализ служит основой для принятия решений о необходимых
мерах контроля и прогнозирования нагрузки на систему здравоохранения.
